```python
import pandas as pd
from pathlib import Path

# Use the same features and feature order used to train the model
features = [
    "category_code",
    "supplier_country_code",
    "retail_price",
    "quantity"
]

# Confirm all required feature columns exist
missing_features = [col for col in features if col not in df.columns]

if missing_features:
    raise ValueError(
        f"Missing feature columns: {missing_features}. "
        "Create these columns before generating predictions."
    )

# Build the full feature matrix: one prediction per df row
X_all = df[features]

# Check for missing values that would prevent valid predictions
print("Missing values in model features:")
print(X_all.isnull().sum())

# Find the probability column that represents Returned = 1
positive_class_index = list(model.classes_).index(1)

# Add the model's predicted probability of return to every row
df = df.copy()

df["return_probability"] = model.predict_proba(X_all)[:, positive_class_index]

# Inspect probabilities before filtering
print("\nReturn-probability summary:")
print(df["return_probability"].describe())

print("\nRows with probability > 0.60:")
print((df["return_probability"] > 0.60).sum())
```

```python
high_risk = df[df['return_probability'] > 0.5]
high_risk_products = high_risk[['product_id','product_name','category','supplier_name','return_probability']]
```

```python
# Export
high_risk_products.to_csv("high_risk_products.csv", encoding="utf-8-sig", index=False)
```

```python
# One risk score per product: average probability across its order-item rows
product_risk = (
    df.groupby(
        ["product_id", "product_name", "category", "supplier_name"],
        as_index=False
    )["return_probability"]
    .mean()
    .sort_values("return_probability", ascending=False)
)

top_10_percent_threshold = product_risk["return_probability"].quantile(0.90)

high_risk_products = product_risk[
    product_risk["return_probability"] >= top_10_percent_threshold
].copy()

# Save to the original CSV-data folder
output_path = Path(
    r"C:\Disk Partition\datax-labs\Project Phase\Phase 1.2\Synthetic data"
) / "high_risk_products.csv"

high_risk_products.to_csv(
    output_path,
    encoding="utf-8-sig",
    index=False
)

print(f"\nSaved file: {output_path}")
print(f"High-risk products exported: {len(high_risk_products)}")
print(high_risk_products.head(10))
```
